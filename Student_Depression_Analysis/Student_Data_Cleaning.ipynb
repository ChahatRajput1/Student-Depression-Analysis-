#Replacing categorical to numerical following columns
Gender,Sleep Duration,Dietary Habits and Have you ever had suicidal thoughts ?

```python
df['Gender']=df['Gender'].replace({"Male":0,"Female":1})
```

```python
df['Sleep Duration']=df['Sleep Duration'].replace({"More than 8 hours":0,"7-8 hours":1,"5-6 hours":2,"Less than 5 hours":3})
```

```python
df['Have you ever had suicidal thoughts ?']=df['Have you ever had suicidal thoughts ?'].replace({"Yes":0,"No":1})
```

```python
df['Family History of Mental Illness']=df['Family History of Mental Illness'].replace({"Yes":0,"No":1})
```

```python
df['Dietary Habits']=df['Dietary Habits'].replace({"Moderate":0,"Healthy":1,"Unhealthy":2})
```
